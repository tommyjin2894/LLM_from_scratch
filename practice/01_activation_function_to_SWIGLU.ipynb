# ReLU

$$ \text{ReLU} = \text{max}(0,x)$$
$$ \text{Sigmoid} = \frac{1}{1+e^{-x}}$$
$$ \text{SiLU,Swish} = x \times \text{Sigmoid(x)} = \frac{x}{1+e^{-x}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 1000)

def relu(x): return np.maximum(0, x)
def sigmoid(x): return 1 / (1 + np.exp(-x))
def silu(x): return x * sigmoid(x)

plt.plot(x, relu(x), label="ReLU")
plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, silu(x), label="SiLU==Swish")
plt.legend()
plt.grid(True)
plt.show()
```

# SwiGLU
$$
\text{GLU}(x,W,V,b,c) = \sigma{(xW + b)} \odot (xV + c)
$$
를 층에 적용 시키면서 간소화
$$
\text{SwiGLU} =\text{SiLU}(W_1 x) \times W_2 x
$$

```python
import numpy as np

def swiglu(x, W1, W2):
    gate = silu(x @ W1)
    hidden = x @ W2
    return gate * hidden

# 테스트
x = np.random.randn(4, 8)   # (batch, dim)
W1 = np.random.randn(8, 16) # (dim, hidden)
W2 = np.random.randn(8, 16)

out = swiglu(x, W1, W2)
print(out.shape)  # (4, 16)
```
